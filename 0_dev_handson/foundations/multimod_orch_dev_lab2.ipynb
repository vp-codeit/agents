```python
request = "For different workflows in Agentic AI few being prompt chaining, routing, parallelisation, orchestrator-worker and Evaluator-Optimiser, I need input prompts for each type that can clearly demonstrate each workflow's application. I need prompts that ask for human inputs and not any set examples as per your assumptions."
request += f"Answer in a dictionary format with the name of workflow and its corresponding descriptive prompt that makes it easy for the code to iterate over."
messages = [{"role":"user","content":request}]
```

```python
model = "gpt-4.1-nano"
openai = OpenAI()

response = openai.chat.completions.create(
    model = model,
    messages= messages)

answer = response.choices[0].message.content
display(Markdown(answer))
```

```python
answer_dict = json.loads(answer)
print(answer_dict)
```

```python
def iterate_nested_dict(d):
    for k,v in d.items():
        if isinstance(v,dict):
            for key in list(v.keys()):
                if(key == "description"):
                    del v[key]
                else:
                    d[k]=v[key]
                    #print(v[key])
    #print(d)
    return d
```

```python
answer_dict = json.loads(answer)
#print(answer_dict)
workflow_prompt_dict = iterate_nested_dict(answer_dict)
print(workflow_prompt_dict)
```

```python
def workflow_demo(workflow_name,model_name):
    request = f"""For showing the workflow demo for the type {workflow_name}, Ask the user for all required inputs step-by-step.
                Do NOT assume any default values.
                Do NOT skip any required inputs."""
    request = request+f"{workflow_prompt_dict[workflow_name]}"

    messages = [{"role":"user","content":request}]
    #model = "gpt-4.1-nano"
    openai = OpenAI()
    while True:
        response = openai.chat.completions.create(
            model = model_name,
            messages= messages)

        answer = response.choices[0].message.content
        display(Markdown(answer))

        messages.append({"role": "assistant", "content": answer})

        user_input = input(f"\n Your input (type 'exit' to stop): ")

        if user_input.lower() == "exit":
            break

        messages.append({"role": "user", "content": user_input})
```

```python
workflow_demo("Evaluator-Optimiser","gpt-4.1-nano")
```
